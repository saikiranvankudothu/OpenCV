```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
height,width = 500,700
blue = (255,127,0)
canvas = np.full((height,width,3),blue,dtype=np.uint8)
plt.imshow(cv2.cvtColor(canvas,cv2.COLOR_BGR2RGB)) 
# plt.imshow(canvas) return orange  layout
```

⚠️ If plt.imshow(canvas) is used without cv2.cvtColor, the BGR values are interpreted as RGB, resulting in a Orange tint instead of the intended blue.

```python
cv2.line(canvas,(50,50),(650,450),(255,255,255),5)

plt.imshow(cv2.cvtColor(canvas,cv2.COLOR_BGR2RGB))
```

### Draw a Rectangle

```python
cv2.rectangle(canvas,(100,100),(300,400),(0,255,255),-1)
plt.imshow(cv2.cvtColor(canvas,cv2.COLOR_BGR2RGB))
```

### Draw a circle

```python
cv2.circle(canvas,(500,150),100,(0,0,255),5)
plt.imshow(cv2.cvtColor(canvas,cv2.COLOR_BGR2RGB))
```

### Draw an Ellipse

```python
cv2.ellipse(canvas,(350,350),(100,50),45,0,360,(0,255,0),3)
plt.imshow(cv2.cvtColor(canvas,cv2.COLOR_BGR2RGB))
```

### Draw a polygon

```python
points = np.array([[150,400],[300,400],[255,300]],np.int32)
new_point = points.reshape((-1,1,2))
print(new_point)
```

```python
cv2.polylines(canvas,[new_point],True,(255,0,255),3)

plt.imshow(cv2.cvtColor(canvas,cv2.COLOR_BGR2RGB))
```
