### Bluring
Bluring is a technique which reduces the sharp transition in an image by averaging pixel values in a local neighborhood



```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
image = cv2.imread("asserts/giraffe-Kenya.png")
```

```python
image_rgb = cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
image_grey = cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
```

```python
kernel_s = 11
average_blur = cv2.blur(image_grey,(kernel_s,kernel_s))

plt.imshow(average_blur,cmap="grey")
```

```python
kernel_s = 13 # should be odd
guassion_blur = cv2.GaussianBlur(image_grey,(kernel_s,kernel_s),0)
plt.imshow(guassion_blur,cmap="gray")
```

```python
median_blur = cv2.medianBlur(image_grey,11)
plt.imshow(median_blur,cmap="gray")
```

```python
plt.figure(figsize=(14, 12))

plt.subplot(2, 2, 1)
plt.title("Original Grayscale")
plt.imshow(image_grey, cmap="gray")
plt.axis('off')

plt.subplot(2, 2, 2)
plt.title("Average Blurred")
plt.imshow(average_blur, cmap="gray")
plt.axis('off')

plt.subplot(2, 2, 3)
plt.title("Gaussian Blurred")
plt.imshow(guassion_blur, cmap="gray")  # also corrected variable name if needed
plt.axis('off')

plt.subplot(2, 2, 4)
plt.title("Median Blurred")
plt.imshow(median_blur, cmap="gray")
plt.axis('off')

```
