A Image filter is a set of rules that we apply to an image to extract or emphasize certain features

    ----------------->Smoothens        
    |
filter ---------------> Sharpens

    |
    -----------------> Enhances

Filter depends on kernel,which is a 3X3 or 5X5 or NXN matrix which determines how filter operates



```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("asserts/img4.jpg",cv2.IMREAD_GRAYSCALE)
plt.imshow(image,cmap='gray')
plt.axis("off")
```

```python
x=-1
horizontal_kernel = np.array([[-1,-1,-1],
                              [x,x,x],
                              [1,1,1]])

vertical_kernel = np.array([[-1,x,1],
                            [-1,x,1],
                            [ -1,x,1]])
                            
diagonal_kernel = np.array([[1,1,x], 
                            [1,x,-1],
                            [x,-1,-1]])

horizontal_filtered_image = cv2.filter2D(image,-1,horizontal_kernel)
vertical_filtered_image = cv2.filter2D(image,-1,vertical_kernel)
diagonal_filtered_image = cv2.filter2D(image,-1,diagonal_kernel)

plt.figure(figsize=(10,10))
plt.subplot(2,2,1)
plt.title("Original image")
plt.imshow(image,cmap="gray")
plt.axis("off")

plt.subplot(2,2,2)
plt.title("Horizontal image")
plt.imshow(horizontal_filtered_image,cmap="gray") 
plt.axis("off")

plt.subplot(2,2,3)
plt.title("Vertical image")
plt.imshow(vertical_filtered_image,cmap="gray") 
plt.axis("off")

plt.subplot(2,2,4)
plt.title("Diagonal image")
plt.imshow(diagonal_filtered_image,cmap="gray") 
plt.axis("off")
```
