## 📘 Introduction to Color Spaces in OpenCV
OpenCV represents images in different color spaces—systems for organizing colors in a consistent way. Understanding color spaces is crucial for tasks like image segmentation, color-based filtering, and feature extraction.

The most commonly used color spaces in OpenCV are:

* BGR (Blue, Green, Red): Default in OpenCV

* RGB (Red, Green, Blue): Common in most image display systems

* Grayscale: Simplifies images to single intensity channel

* HSV (Hue, Saturation, Value): Useful for color detection and manipulation

* LAB (Lightness, A (green–red), B (blue–yellow)): Mimics human vision perception

### Step 1: Import OpenCV

```python
import cv2
```

### Step 2: Load the BGR Image
OpenCV reads images in BGR format by default.

```python
bgr_image = cv2.imread("./img1.png")
```

### Step 3: Convert to Grayscale
Grayscale removes color and keeps only intensity information, reducing computational complexity.

```python
# convert to grayscale image
gray_image = cv2.cvtColor(bgr_image,cv2.COLOR_BGR2GRAY)
cv2.imwrite("asserts/gray_img1.png",gray_image)
```

### Step 4: Convert to HSV
HSV separates color information (Hue) from intensity (Value), making it great for object tracking and segmentation.

```python
# convert to hsv image
hsv_image = cv2.cvtColor(bgr_image,cv2.COLOR_BGR2HSV)
cv2.imwrite("asserts/hsv_img1.png",hsv_image)
```

### Step 5: Convert to LAB
LAB color space is designed to approximate human vision. It is perceptually uniform, meaning equal changes in value correspond to equal changes in perceived color.

```python
# convert to lab image
lab_image = cv2.cvtColor(bgr_image,cv2.COLOR_BGR2LAB)
cv2.imwrite("asserts/lab_img1.png",lab_image)
```

### Step 6: Convert to RGB
Although OpenCV loads images as BGR, you may want to convert them to RGB for correct color display in libraries like matplotlib.

```python
# convert to rgb
rgb_image = cv2.cvtColor(bgr_image,cv2.COLOR_BGR2RGB)
cv2.imwrite("asserts/rgb_img1.png",rgb_image)
```
